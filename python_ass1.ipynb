```python
#3.2.Display dictionary keys, values, and subject with highest marks
subjects_marks = {
    "Math": 95,
    "Science": 88,
    "English": 78,
    "History": 92,
    "Computer": 98
}

print("Subject Marks Dictionary:", subjects_marks)

# Display all keys (subject names)
print("\nAll Subject Names (Keys):")
for subject in subjects_marks.keys():
    print(subject)

# Display all values (marks)
print("\nAll Marks (Values):")
for marks in subjects_marks.values():
    print(marks)

# Find the subject with the highest marks
highest_marks = 0
highest_subject = ""

for subject, marks in subjects_marks.items():
    if marks > highest_marks:
        highest_marks = marks
        highest_subject = subject

print(f"\nSubject with the Highest Marks: {highest_subject} ({highest_marks})")
```

```python
#8.1.SET
set1={10,20,30,40,70}
set2={90,80,70,60,50}
print("set1:",set1)
print("set2:",set2)
union=set1.union(set2)
print("union:",union)
intersection=set1.intersection(set2)
print("intersection:",intersection)
difference=set1.difference(set2)
print("difference(set1-set2):",difference)
difference=set2.difference(set1)
print("difference(set2-set1):",difference)
print("symmetric difference",set1.symmetric_difference(set2))
```

```python
#8.2. Remove duplicates from a list using a set
original_list = [10, 20, 30, 40, 50, 50, 40, 30, 90, 20, 60]
print("Original list:", original_list)

unique_elements_set = set(original_list)
print("Unique elements (as a set):", unique_elements_set)

# If you need the result back as a list:
unique_elements_list = __builtins__.list(unique_elements_set)
print("Unique elements (as a list):", unique_elements_list)
print("Number of unique elements:", len(unique_elements_list))
```
