Question 5 — NetworkX
A school wants to visualize the friendship connections among a small group of students to understand
how they are socially linked.
Consider the following friendship connections:
Student A Student B
Alice Bob
Alice Charlie
Bob David
Charlie David
David Eva
Eva Frank
Charlie Frank
Task:
Using NetworkX, write a Python program to:
1. Create a graph representing the friendship network using the given connections.
2. Visualize the network graph with proper node labels, node colors, and edge styling.
3. Display the total number of nodes and edges in the graph.
4. Compute and print the degree of each student (number of friends).
5. Identify the student(s) with the highest number of friends.
6. Check and display whether a direct friendship exists between "Alice" and "Eva".
7. Find and display the shortest path (and its length) between "Alice" and "Frank".
8. Write 1–2 observations about the overall structure of the friendship network.
Expected Outcome:
A clearly labelled friendship network graph along with basic structural information such as node/edge
counts, degrees, and shortest paths between students.

```python
import networkx as nx
import matplotlib.pyplot as plt

# 1. Create the friendship graph
G = nx.Graph()

# Given friendship connections
friendships = [
    ("Alice", "Bob"),
    ("Alice", "Charlie"),
    ("Bob", "David"),
    ("Charlie", "David"),
    ("David", "Eva"),
    ("Eva", "Frank"),
    ("Charlie", "Frank")
]

G.add_edges_from(friendships)

# 2. Visualize the network
plt.figure(figsize=(8, 6))

# Generate positions for the nodes
pos = nx.spring_layout(G, seed=42)

# Draw nodes
nx.draw_networkx_nodes(
    G,
    pos,
    node_color="lightyellow",
    node_size=2000,
    edgecolors="orange"
)

# Draw edges
nx.draw_networkx_edges(
    G,
    pos,
    edge_color="lightpink",
    width=2
)

# Draw labels
nx.draw_networkx_labels(
    G,
    pos,
    font_size=11,
    font_weight="bold"
)

plt.title("Student Friendship Network")
plt.axis("off")
plt.show()

# 3. Display total number of nodes and edges
print("Total number of students (nodes):", G.number_of_nodes())
print("Total number of friendships (edges):", G.number_of_edges())

# 4. Compute and print degree of each student
print("\nDegree of each student:")
for student, degree in G.degree():
    print(f"{student}: {degree}")

# 5. Identify student(s) with the highest number of friends
degrees = dict(G.degree())
max_degree = max(degrees.values())

most_connected = [
    student for student, degree in degrees.items()
    if degree == max_degree
]

print("\nStudent(s) with the highest number of friends:")
print(", ".join(most_connected))
print("Number of friends:", max_degree)

# 6. Check whether Alice and Eva are directly connected
if G.has_edge("Alice", "Eva"):
    print("\nDirect friendship between Alice and Eva: Yes")
else:
    print("\nDirect friendship between Alice and Eva: No")

# 7. Find the shortest path between Alice and Frank
shortest_path = nx.shortest_path(G, source="Alice", target="Frank")
path_length = nx.shortest_path_length(G, source="Alice", target="Frank")

print("\nShortest path from Alice to Frank:")
print(" -> ".join(shortest_path))
print("Path length:", path_length)

# 8. Observations
print("\nObservations:")
print("- Charlie and David are the most connected students, with 3 friends each.")
print("- The network is connected, and Alice can reach Frank through multiple friendship paths.")
```

Question 6 — NetworkX
A telecom company wants to analyze its network of towers to understand connectivity and identify
important hubs.
Consider the following tower connections (edges represent direct signal links):
Tower A Tower B
T1 T2
T1 T3
T2 T4
T3 T4
T4 T5
T5 T6
T3 T6
T6 T7
T2 T7
Task:
Using NetworkX, write a Python program to:
1. Create a graph representing the tower network using the given edges.
2. Visualize the network graph with proper node labels, node colors, and edge styling.
3. Calculate and display the total number of nodes and edges.
4. Compute the degree of each node and identify the tower(s) with the highest degree (most
connections).
5. Calculate and display the shortest path (and its length) between T1 and T7.
6. Compute and display the following centrality measures for all nodes: Degree Centrality,
Betweenness Centrality, and Closeness Centrality.
7. Identify whether the network is fully connected. If not, identify the number of connected components.
8. Highlight the tower with the highest betweenness centrality on the graph visualization (e.g., using a
different color/size).
9. Write 2–3 observations about which towers are most critical to the network based on the computed
metrics.
Expected Outcome:
A visualized network graph along with computed structural metrics that help identify the most important
and best-connected towers in the network.

```python
import networkx as nx
import matplotlib.pyplot as plt

# ---------------------------------------------------------
# 1. Create the graph
# ---------------------------------------------------------

G = nx.Graph()

# Given tower connections
edges = [
    ("T1", "T2"),
    ("T1", "T3"),
    ("T2", "T4"),
    ("T3", "T4"),
    ("T4", "T5"),
    ("T5", "T6"),
    ("T3", "T6"),
    ("T6", "T7"),
    ("T2", "T7")
]

G.add_edges_from(edges)

# ---------------------------------------------------------
# 2. Calculate basic graph information
# ---------------------------------------------------------

print("========== NETWORK INFORMATION ==========")
print("Total number of nodes:", G.number_of_nodes())
print("Total number of edges:", G.number_of_edges())

# ---------------------------------------------------------
# 3. Degree of each node
# ---------------------------------------------------------

degrees = dict(G.degree())

print("\n========== DEGREE OF EACH TOWER ==========")
for node, degree in degrees.items():
    print(f"{node}: {degree}")

max_degree = max(degrees.values())
highest_degree_nodes = [
    node for node, degree in degrees.items()
    if degree == max_degree
]

print("\nTower(s) with highest degree:",
      ", ".join(highest_degree_nodes))
print("Highest degree:", max_degree)

# ---------------------------------------------------------
# 4. Shortest path from T1 to T7
# ---------------------------------------------------------

shortest_path = nx.shortest_path(G, source="T1", target="T7")
shortest_path_length = nx.shortest_path_length(
    G, source="T1", target="T7"
)

print("\n========== SHORTEST PATH ==========")
print("Shortest path from T1 to T7:",
      " -> ".join(shortest_path))
print("Path length:", shortest_path_length)

# ---------------------------------------------------------
# 5. Centrality measures
# ---------------------------------------------------------

degree_centrality = nx.degree_centrality(G)
betweenness_centrality = nx.betweenness_centrality(G)
closeness_centrality = nx.closeness_centrality(G)

print("\n========== CENTRALITY MEASURES ==========")

print("\nDegree Centrality:")
for node, value in degree_centrality.items():
    print(f"{node}: {value:.4f}")

print("\nBetweenness Centrality:")
for node, value in betweenness_centrality.items():
    print(f"{node}: {value:.4f}")

print("\nCloseness Centrality:")
for node, value in closeness_centrality.items():
    print(f"{node}: {value:.4f}")

# Tower with highest betweenness centrality
max_betweenness = max(betweenness_centrality.values())

highest_betweenness_nodes = [
    node for node, value in betweenness_centrality.items()
    if value == max_betweenness
]

print("\nTower(s) with highest betweenness centrality:",
      ", ".join(highest_betweenness_nodes))

# ---------------------------------------------------------
# 6. Check connectivity
# ---------------------------------------------------------

connected = nx.is_connected(G)
components = list(nx.connected_components(G))

print("\n========== CONNECTIVITY ==========")

if connected:
    print("The network is fully connected.")
else:
    print("The network is NOT fully connected.")

print("Number of connected components:", len(components))

# ---------------------------------------------------------
# 7. Visualize the network
# ---------------------------------------------------------

# Spring layout for a clear network visualization
pos = nx.spring_layout(G, seed=42)

# Normal nodes
normal_nodes = [
    node for node in G.nodes()
    if node not in highest_betweenness_nodes
]

# Draw edges
nx.draw_networkx_edges(
    G,
    pos,
    edge_color="gray",
    width=2,
    style="solid"
)

# Draw normal nodes
nx.draw_networkx_nodes(
    G,
    pos,
    nodelist=normal_nodes,
    node_color="lightyellow",
    node_size=1000,
    edgecolors="orange"
)

# Highlight highest-betweenness tower(s)
nx.draw_networkx_nodes(
    G,
    pos,
    nodelist=highest_betweenness_nodes,
    node_color="pink",
    node_size=1600,
    edgecolors="black"
)

# Node labels
nx.draw_networkx_labels(
    G,
    pos,
    font_size=11,
    font_weight="bold"
)

plt.title("Telecom Tower Network")
plt.axis("off")
plt.tight_layout()
plt.show()

# ---------------------------------------------------------
# 8. Observations
# ---------------------------------------------------------

print("\n========== OBSERVATIONS ==========")
print(
    f"1. {', '.join(highest_degree_nodes)} has the highest degree "
    f"({max_degree}), making it one of the best-connected towers."
)

print(
    f"2. {', '.join(highest_betweenness_nodes)} has the highest "
    f"betweenness centrality ({max_betweenness:.4f}), indicating "
    "that it plays an important role in connecting different parts "
    "of the network."
)

print(
    "3. Since the network is fully connected, every tower can reach "
    "every other tower through one or more signal links."
)
```
